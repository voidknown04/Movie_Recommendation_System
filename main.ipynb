```python
movies_df = movies.merge(credits, left_on='id', right_on='movie_id', how='inner')
print(movies_df.columns)
print(movies_df.shape)
```

```python
# Numerical Features (Useful for Recommendation):
# movie_id
# budget 
# revenue
# runtime 
# keywords
# original_title
# popularity
# vote_average
# vote_count

# Categorical/Text Features (Useful for Content-Based Filtering):
# title_x
# genres 
# keywords
# original_language
# overview 
# cast 
# crew
```

```python
df = movies_df[['genres','movie_id', 'budget', 'revenue', 'runtime', 'keywords', 'original_title', 'popularity', 'vote_average', 'vote_count']]
```

```python
print(df.shape)
df.info()
```

```python
print(df.isnull().sum())
df.dropna(inplace=True)
df.isnull().sum()
print(df.duplicated().sum())
```

```python
import ast  # To parse string representations of lists

def convert(text):
    L = []
    for i in ast.literal_eval(text):
        L.append(i['name']) 
    return L 

```

```python
df['genres'] = df['genres'].apply(convert)
df.head()
```

```python
df['keywords'] = df['keywords'].apply(convert)
df['cast'] = df['cast'].apply(convert)
df['crew'] = df['crew'].apply(convert)
```
